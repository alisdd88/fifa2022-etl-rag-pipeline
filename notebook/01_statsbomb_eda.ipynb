````python
events_columns = [
    "match_id",
    "id",
    "index",
    "period",
    "timestamp",
    "duration",
    "type",
    "player_id",
    "foul_committed_penalty",
    "foul_committed_offensive",
    "foul_committed_type",
    "foul_committed_advantage",
    "bad_behaviour_card",
    "foul_committed_card",
    "shot_outcome",
    "shot_saved_to_post",
    "shot_type",
    "substitution_replacement",
    "substitution_replacement_id",
    "substitution_outcome",
    "substitution_outcome_id",
]
````

````python
events_df_processed = events_df[events_columns]
````

````python
events_df_processed.head()
````

````python
events_df_processed['type'].unique()
````

| `type` | Why keep it | Important fields |
|---|---|---|
| `Foul Committed` | Every foul committed by a player | `foul_committed_type`, `foul_committed_card`, `foul_committed_penalty`, `foul_committed_offensive` |
| `Shot` | Goals, including penalty goals | `shot_outcome`, `shot_type`, `shot_body_part`, `shot_statsbomb_xg`, etc. |
| `Bad Behaviour` | Cards given outside a normal foul event | `bad_behaviour_card` |
| `Substitution` | Player substitutions | `player`, `player_id`, `substitution_replacement`, `substitution_replacement_id`, `substitution_outcome` |

````python
event_types = ["Foul committed", "Shot", "Bad behaviour", "Substitution"]
filtered_events = events_df_processed[events_df_processed["type"].isin(event_types)]
````

````python
filtered_events.head()
````

````python
substitutions = filtered_events[filtered_events["type"] == "Substitution"]
````

````python
substitutions.sort_values(by='player_id')
````

| Raw condition | Processed `event_type` | Player | Metadata |
|---|---|---|---|
| `Foul Committed` + card exists | `card` | foul player | `card_reason=foul`, `card_type=...` |
| `Bad Behaviour` + card exists | `card` | player | `card_reason=bad_behaviour`, `card_type=...` |
| `Shot` + `shot_outcome == Goal` | `goal` | shooter | `shot_type=open_play/penalty/...` |
| `Substitution` | `substitution` | outgoing player | `substitution_type=OUT`, `substitution_ref_id=...` |
| same substitution | `substitution` | incoming player | `substitution_type=IN`, `substitution_ref_id=...` |
